### **2. DESCRIPCIÓN DE LAS COLUMNAS:**

1. State Code: "Código de Estado".
Es el código "FIPS" del Estado en el que se ubica el monitor de medición.

2. County Code: "Código de Condado".
Es el código "FIPS" del Condado en el que se ubica el monitor de medición.

3. Site Num: "Número de sitio".
Número único dentro del Condado que identifica el sitio.

4. Parameter Code: "Código de parámetro".
Código "AQS" correspondiente al parámetro medido por el monitor.

5. POC: "Código de Ocurrencia del Parámetro".
Se utiliza para identificar de forma unívoca un monitor si hay más de un dispositivo midiendo el mismo contaminante en un sitio

6. Latitude: "Latitud".
Latitud de la ubicación del monitor, medida en grados.

7. Longitude: "Longitud".
Longitud de la ubicación del monitor, medida en grados.

8. Datum: "Datum".
Datum asociado a las medidas de latitud y longitud.

9. Parameter Name: "Nombre del parámetro".
Nombre o descripción asignado en "AQS" al parámetro medido por el monitor. Los parámetros pueden ser contaminantes o no contaminantes.

10. Sample Duration: "Duración de la muestra".
Es el tiempo durante el cual el aire pasa por el monitor antes de ser analizado (medido). Representa un período de promedio en la atmósfera (por ejemplo, una muestra de 24 horas toma aire ambiente sobre un filtro de recolección durante 24 horas seguidas). En el caso de monitores continuos, puede representar el tiempo promedio de muchas muestras (por ejemplo, un valor de 1 hora puede ser el promedio de cuatro muestras de 1 minuto tomadas durante cada cuarto de hora).

11. Pollutant Standard: "Standard de polución".
Descripción de las normas de calidad de aire utilizadas para recopilar estadísticas.

12. Metric Used: "Métrica utilizada".
Es la base métrica empleada en el cálculo de las estadísticas agregadas que se presentan en el resto de la fila. Por ejemplo: si se trata del "Máximo Diario", el valor de la columna "Media" es la media de los máximos diarios.

13. Method Name: "Nombre del método".
Breve descripción de los procesos, equipos y protocolos utilizados para la recolección y medición de la muestra.

14. Year: "Año".
El año que representan los datos del resumen anual.

15. Units of Measure: "Unidades de medida".
La unidad de medida del parámetro. "QAD" siempre devuelve los datos en las unidades estándar del parámetro.

16. Event Type: "Tipo de evento".
Indica si los datos medidos durante eventos excepcionales se incluyen en el resumen. Un incendio forestal es un ejemplo de evento excepcional; afecta la calidad del aire, pero la agencia local no tiene control sobre él.
    * a. "Sin eventos" significa que no se registraron eventos.
    * b. "Eventos incluidos" significa que se registraron eventos y sus datos se incluyen en el resumen.
    * c. "Eventos excluidos" significa que se registraron eventos, pero sus datos se excluyen del resumen.
    * d. "Eventos concurrentes excluidos" significa que se registraron eventos, pero sólo se eliminan del resumen las exclusiones concurrentes de la "EPA". Si se registró un evento para el parámetro en cuestión, los datos tendrán varios registros para cada monitor.

17. Observation Count: "Recuento de observaciones".
Es el número de observaciones (muestras) tomadas durante el año.

18. Observation Percent: "Porcentaje de observación".
Porcentaje que representa el número de observaciones realizadas con respecto al número previsto para el año. Este cálculo se aplica únicamente a los monitores que requieren mediciones (sólo para algunos parámetros).

19. Completeness Indicator: "Indicador de integridad".
Indica si el monitor ha cumplido los criterios reglamentarios de integridad de datos para obtener datos resumidos válidos durante el año.

    * "Y" significa "si".
    * "N" significa "no" o que no existen criterios reglamentarios de integridad para el parámetro.

20. Valid Day Count: "Conteo de días válidos".
El número de días durante el año en los que se cumplieron los criterios de monitoreo diario, si el cálculo de los resúmenes se basa en días válidos.

21. Required Day Count: "Número de días requeridos".
El número de días durante el año en que el monitor estaba programado para tomar muestras si se requieren mediciones.

22. Exceptional Data Count: "Recuento de datos excepcionales".
Número de puntos de datos en el conjunto de datos anual afectados por eventos excepcionales de calidad del aire (elementos fuera de lo normal que afectan la calidad del aire).

23. Null Data Count: "Recuento de datos nulos".
Número de muestras programadas en las que no se recopilaron datos y se informó el motivo de la falta de datos.

24. Primary Exceedance Count: "Recuento de excedencias primarias".
Número de muestras tomadas durante el año que superaron el estándar primario de calidad del aire.

25. Secondary Exceedance Count: "Recuento de excedencias secundarias".
Número de muestras tomadas durante el año que superaron el estándar secundario de calidad del aire.

26. Certification Indicator: "Indicador de certificación".
Indica si el remitente ha certificado la integridad y exactitud de la información del resumen anual.

    * "Certificado" significa que el remitente ha certificado los datos (fecha límite: 1 de mayo del año siguiente a la recopilación).
    * "No se requiere certificación" significa que el parámetro no requiere certificación o que aún no ha vencido el plazo.
    * "Sin certificar (vencido)" significa que la certificación es obligatoria, pero está vencida.
    * "Solicitado, pero aún no aprobado" significa que el remitente ha completado el proceso, pero la EPA aún no ha certificado los datos.
    * "Solicitado, pero denegado" significa que el remitente ha completado el proceso, pero la EPA ha denegado la solicitud por causa justificada.
    * "Certificado, pero datos modificados" significa que los datos fuerton certificados, pero se reemplazaron y el proceso no se ha repetido.

27. Num Obs Below MDL: "Número de observaciones debajo de MDL", siendo MDL "Límite de Detección del Método".
Número de muestras analizadas durante el año que estuvieron por debajo del MDL del instrumento de monitoreo. En ocasiones, estos valores se sustituyen por la mitad del MDL en los cálculos de resumen.

28. Arithmetic Mean: "Media aritmética".
Es el promedio aritmético de la medición del contaminante durante un intervalo de muestreo.

29. Arithmetic Standard Dev: "Desviación estándar aritmética".
La desviación estándar con respecto a la media de los valores del año.

30. 1st Max Value: "Primer valor máximo".
El valor más alto del año.

31. 1st Max DateTime: "Fecha del Primer valor máximo".
Fecha y hora (en un reloj de 24 horas) en que se tomó el valor más alto del año (columna anterior).

32. 2nd Max Value: "Segundo valor máximo".
El segundo valor más alto del año.

33. 2nd Max DateTime: "Fecha del Segundo valor máximo".
Fecha y hora (en un reloj de 24 horas) en que se tomó el segundo valor más alto del año (columna anterior).

34. 3rd Max Value: "Tercer valor máximo".
El tercer valor más alto del año.

35. 3rd Max DateTime: "Fecha del Tercer valor máximo".
Fecha y hora (en un reloj de 24 horas) en que se tomó el tercer valor más alto del año (columna anterior).

36. 4th Max Value: "Cuarto valor máximo".
El cuarto valor más alto del año.

37. 4th Max DateTime: "Fecha del Cuarto valor máximo".
Fecha y hora (en un reloj de 24 horas) en que se tomó el cuarto valor más alto del año (columna anterior).

38. 1st Max Non Overlapping Value: "Primer valor máximo no superpuesto".
Para promedios de "CO" de 8 horas, el valor más alto del año.

39. 1st NO Max DateTime: "Fecha del Primer valor máximo no superpuesto".
Fecha y hora (en un reloj de 24 horas) en que se tomó el primer valor máximo no superpuesto del año (columna anterior).

40. 2nd Max Non Overlapping Value: "Segundo valor máximo no superpuesto".
Para promedios de "CO" de 8 horas, el segundo valor más alto del año que no comparte ninguna hora con el período de 8 horas del "Primer valor máximo no superpuesto".

41. 2nd NO Max DateTime: "Fecha del Segundo valor máximo no superpuesto".
Fecha y hora (en un reloj de 24 horas) en que se tomó el segundo valor máximo no superpuesto del año (columna anterior).

42. 99th Percentile: "Percentil 99".
Valor de este monitor para el cual el 99% del resto de los valores medidos durante el año son iguales o menores.

43. 98th Percentile: "Percentil 98".
Valor de este monitor para el cual el 98% del resto de los valores medidos durante el año son iguales o menores.

44. 95th Percentile: "Percentil 95".
Valor de este monitor para el cual el 95% del resto de los valores medidos durante el año son iguales o menores.

45. 90th Percentile: "Percentil 90".
Valor de este monitor para el cual el 90% del resto de los valores medidos durante el año son iguales o menores.

46. 75th Percentile: "Percentil 75".
Valor de este monitor para el cual el 75% del resto de los valores medidos durante el año son iguales o menores.

47. 50th Percentile: "Percentil 50".
Valor de este monitor para el cual el 50% del resto de los valores medidos durante el año son iguales o menores (es decir, la mediana).

48. 10th Percentile: "Percentil 10".
Valor de este monitor para el cual el 10% del resto de los valores medidos durante el año son iguales o menores.

49. Local Site Name: "Nombre del sitio local".
Nombre del sitio (si lo hay) proporcionado por la agencia estatal que opera el control de la contaminación del aire.

50. Address: "Dirección".
Dirección aproximada de la ubicación del sitio de monitoreo.

51. State Name: "Nombre del Estado".
Nombre del Estado donde se encuentra el sitio de monitoreo.

52. County Name: "Nombre del Condado".
Nombre del Condado donde se encuentra el sitio de monitoreo.

53. City Name: "Nombre de la Ciudad".
El nombre de la ciudad donde se ubica el sitio de monitoreo. Representa los límites legales incorporados de las ciudades y no las áreas urbanas.

54. CBSA Name: "Nombre CBSA", siendo CBSA "Core-Based Statistical Area" (Área Estadística Basada en Núcleos).
Nombre del área metropolitana donde se encuentra el sitio de monitoreo.

55. Date of Last Change: "Fecha de última modificación".
Fecha en que se actualizaron por última vez los valores númericos de este registro en el sistema de datos "AQS".

### Estudiaremos a la siguiente variable: `arithmetic_mean`

Repasando su función, es el promedio aritmético de la medición del contaminante durante un intervalo de muestreo.

Ejemplo:
Si un sensor mide ozono cada minuto durante 8 horas, el dataset reporta el promedio de esos valores.



#### Haremos el análisis con los siguientes filtros:
| parameter_name | state_name | units_of_measure | year|
| :--- | :---: |:---: |:---: |
| 1. Ozone | California | Parts per million |Sin filtro


```python
df_filt = df_filtrado_name[
    (df_filtrado_name["parameter_name"] == "Ozone") &
    (df_filtrado_name["state_name"] == "California") &
    (df_filtrado_name["units_of_measure"] == "Parts per million")
]

df_filt = df_filt.reset_index(drop=True)
df_filt.head()
```

##### Media

```python
media = df_filt["arithmetic_mean"].mean()
media
```

##### Mediana

```python
mediana = df_filt["arithmetic_mean"].median()
mediana
```

##### Moda

```python
moda = df_filt.arithmetic_mean.mode().iloc[0]
moda
```

##### Cuartiles

```python
df_filt["arithmetic_mean"].quantile([0.25, 0.5, 0.75])
```

##### Boxplot

```python
# 1. Definimos una figura y UN solo eje (subplot).
fig, ax = plt.subplots(figsize=(15, 7))

#2. Graficar el boxplot para 'df_filt' en el único eje 'ax'
sns.boxplot(data=df_filt, y='arithmetic_mean', ax=ax, color='skyblue')
ax.set_title('Boxplot de los valores promedios de Ozono en California', fontsize=16)

#3. Agregamos un título
ax.set_ylabel('Parts per million', fontsize=16)
plt.tight_layout() # Sin argumentos rect para un layout más simple

#4. Mostramos el gráfico.
plt.show()
```

###### Podriamos pensar que esos tres valores lejanos que se observan pueden llegar a ser outliers "erróneos". Pero también es cierto que en datos ambientales, especiamente en contaminantes como el ozono, nos interesa estudiar con mayor detenimiento a esos picos, ya que podrian representar eventos de alta contaminación real.

###### Graficaremos nuevamente el mismo Boxplot, agregando etiquetas a sus tres valores que sobresalen:

```python
plt.figure(figsize=(4,7))
plt.boxplot(df_filt['arithmetic_mean'])

# valores top 3
top3 = df_filt['arithmetic_mean'].nlargest(3)

for val in top3:
    plt.text(1.05, val, f"{val:.3f}", fontsize=10)

plt.show()
```

##### Histograma

```python
fig, ax = plt.subplots(figsize=(12, 6))
sns.histplot(data=df_filt, x='arithmetic_mean', kde=True, bins=30, ax=ax, color='skyblue')
ax.set_title ('Histograma de los valores promedios de Ozono en California', fontsize=16)
plt.xlabel("Concentración (arithmetic_mean)")
plt.ylabel("Frecuencia")
plt.title("Histograma de valores promedios de ozono en California")



# --- Dibujar Media, Mediana y Moda en el gráfico ---
# Usamos ax.axvline para dibujar líneas verticales.
ax.axvline(media, color='red', linestyle='dashed', linewidth=2, label=f'Media: {media:,.2f}')
ax.axvline(mediana, color='green', linestyle='dashed', linewidth=2, label=f'Mediana: {mediana:,.2f}')
ax.axvline(moda, color='purple', linestyle='dashed', linewidth=2, label=f'Moda: {moda:,.2f}')


# 4. Añadimos la leyenda para identificar las líneas.
ax.legend()

plt.show()

# Imprimimos los valores calculados
print(f"Media: {media:,.2f}")
print(f"Mediana: {mediana:,.2f}")
print(f"Moda: {moda:,.2f}" )
```

La mediana y la media son muy similares, indicando una distribución equilibrada. Se observan algunos días con valores más altos, evidenciados en la cola derecha del histograma, pero la mayor parte de las mediciones se mantiene dentro de un rango acotado, indicando una variabilidad moderada

## Aplicación de tabla pivoteada

```python

# ======================================
# 1) Crear tabla pivoteada
# ======================================

contaminantes = [
    "PM2.5 - Local Conditions",
    "PM10 Total 0-10um STP",
    "Ozone",
    "Carbon monoxide",
    "Sulfur dioxide"
]

df_pivot = df_filtrado_name.pivot_table(
    index=["state_code", "county_code", "site_num", "year"],
    columns="parameter_name",
    values="arithmetic_mean"
).reset_index()

# Filtrar solo las columnas relevantes
df_pivot = df_pivot[["state_code", "year"] + contaminantes]

# ======================================
# 2) Limpiar nulos (opcional)
# ======================================
df_pivot_clean = df_pivot.dropna(subset=contaminantes)

print("Cantidad de datos antes:", len(df_pivot))
print("Cantidad de datos después de limpiar nulos:", len(df_pivot_clean))

# ======================================
# 3) Scatterplot entre dos contaminantes
# ======================================

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df_pivot_clean,
    x="PM2.5 - Local Conditions",
    y="PM10 Total 0-10um STP",
    alpha=0.4
)
plt.title("Relación entre PM2.5 y PM10 en la Zona Oeste")
plt.xlabel("PM2.5 [µg/m³]")
plt.ylabel("PM10 [µg/m³]")
plt.show()

# ======================================
# 4) Matriz de correlación
# ======================================

df_corr = df_pivot_clean[contaminantes].corr()

plt.figure(figsize=(8,6))
sns.heatmap(df_corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matriz de Correlación entre Contaminantes")
plt.show()

```

Conclusiones:
- PM2.5 y PM10 suelen correlacionarse porque ambas partículas provienen de fuentes similares
  (combustión, polvo suspendido, tráfico, incendios forestales).

- Ozone no suele correlacionar con partículas porque tiene un comportamiento diferente,
  depende de radiación solar y reacciones químicas en la atmósfera.

- SO2 y CO dependen de fuentes puntuales (industria, combustibles) y muchas veces aparecen más desconectados.

## Aplicación de binning

### Se generó una nueva variable categórica AQI_Category mediante Binning. La variable arithmetic_mean (continua) se discretizó usando los rangos oficiales EPA para PM2.5. Este proceso convierte niveles numéricos de contaminación en categorías cualitativas (Bueno, Moderado, Insalubre, etc.), facilitando la interpretación, el análisis descriptivo y la visualización   

| **Categoría AQI**          | **Rango PM2.5 (µg/m³)** | **Significado**                        |
| -------------------------- | ----------------------- | -------------------------------------- |
| Bueno                      | 0.0 – 12.0              | Calidad del aire óptima                |
| Moderado                   | 12.1 – 35.4             | Puede afectar a personas muy sensibles |
| Insalubre grupos sensibles | 35.5 – 55.4             | Riesgo en niños, ancianos, asmáticos   |
| Insalubre                  | 55.5 – 150.4            | Riesgo para todos                      |
| Muy insalubre              | 150.5 – 250.4           | Riesgo alto                            |
| Peligroso                  | 250.5 – más             | Evitar estar afuera                    |


```python
# 1. Filtrar solo PM2.5
df_pm25 = df_mejorado[df_mejorado["parameter_name"] == "PM2.5 - Local Conditions"].copy()

# 2. Definir rangos (bins) y etiquetas EPA
bins = [0, 12.0, 35.4, 55.4, 150.4, 250.4, 500]
labels = [
    "Bueno",
    "Moderado",
    "Insalubre grupos sensibles",
    "Insalubre",
    "Muy insalubre",
    "Peligroso"
]

# 3. Crear columna de categoría AQI
df_pm25["pm25_aqi_cat"] = pd.cut(
    df_pm25["arithmetic_mean"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```
